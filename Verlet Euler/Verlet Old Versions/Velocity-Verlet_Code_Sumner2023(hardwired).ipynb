```python
import math
import matplotlib.pyplot as plt
pi=math.pi
time=[]
x_t=[]
v_t=[]
a_t=[]
x_t_exact=[]
error=[]
E=[]
def plotgraph(xarr,yarr,lx,ly):#Takes two arrays and, using the first array's values as x coordinates and the second one's for y coordinates, plots them, and then proceeding to label the x axis as whatever lx is and labels the y axis with the ly argument
  plt.plot(xarr,yarr)
  plt.xlabel(lx)
  plt.ylabel(ly)
def avg(var1, var2):
  p=((var1+var2))
  return p/2
def pr(formula):
  print(formula)
def _2(base_number):
  return (base_number*base_number)
def abs(var):#absolute value function
  var2=_2(var)
  return math.sqrt(var2)
  
  
deltat=.01
deltaT=deltat*pi
xt=1 #Original X value
vt=0 #Original velocity
at=-1 #Original acceleration
error1=0 #We start with no error
E1=.5 #Total energy
k=1 #<--K, or Spring Stiffness *Note: As K increases, so does error
m=1 #<--m, or Mass of Object attached to spring *Note: As M decreases, error increases
x_t.append(xt) #Putting above values into respective arrays
v_t.append(vt)
a_t.append(at)
error.append(error1)
E.append(E1)

def  enmrte(times): #calculates and writes down the different formulas (Main function)
  t1=1 #Creating an index for each variable
  xt1=0
  vt1=0
  at1=0
  xte1=1
  er1=0
  e1=0
  time.append(0)
  x_t_exact.append(1)
  o=0
  for i in range(times-1):
    o=t1*deltat*pi #T
    time.append(o)
    t1+=1
    
    
    o=math.cos(xte1*.01*math.pi)#formula for x(t) exact
    x_t_exact.append(o)        
    xte1+=1

    o=x_t[len(x_t)-1]+v_t[len(x_t)-1]*deltaT+.5*a_t[len(x_t)-1]*_2(deltaT) #x(t)
    x_t.append(o)
    xt1+=1

    o=(-1*k*x_t[len(x_t)-2])/m #a(t)
    a_t.append(o)
    at1+=1

    o=v_t[len(v_t)-1]+.5*(a_t[len(v_t)-1]+a_t[len(a_t)-1]) #v(t)
    v_t.append(o)
    vt1+=1
    
    o=.5*m*_2(v_t[len(v_t)-1])+.5*k*_2(x_t[len(x_t)-1]) #Energy
    E.append(o)
    e1+=1
  
    o=abs(x_t_exact[len(x_t_exact)-1])-abs(x_t[len(x_t)-1])#error
    error.append(o)
    er1+=1
    
   
    
  pr("Delta T/Pi")    #this section prints arrays and their titles
  pr(deltat)
  pr("T")
  pr(time)
  pr("x(t)")
  pr(x_t)
  pr("a(t)")
  pr(a_t)
  pr("x(t) exact")
  pr(x_t_exact)
  pr("error")
  pr(error)
enmrte(100)#executes main  function
```
